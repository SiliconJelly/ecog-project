# 해커톤: 완료된 whitening 비교 결과

이 파일은 **완료된 실험의 결과를 읽는 노트북**입니다. 표와 그림을 바로 볼 수 있습니다.

분석은 `ecog` Python 환경에서 실행했고, 측정값은 `results/whitening/comparison_metrics.json`에 저장되어 있습니다.

| 검사 | 팀원의 기존 방법 | Whitening 추가 |
| --- | --- | --- |
| 가위·바위·보 분류 정확도 | 93.3% ± 7.4% | **96.9% ± 5.7%** |
| 바위·가위만 구분한 정확도 | 93.5% ± 9.4% | 96.7% ± 7.1% |
| 정답을 섞은 대조 검사 | 33.8% | 33.3% |

**같은 평가 조건에서 세 동작의 평균 정확도가 약 3.6%p 높아졌습니다.** `±`는 각 시험 묶음의 점수가 얼마나 달랐는지 나타내는 표준편차입니다.

## 무엇을 했나

1. 팀원의 `classify.py`를 실행해 기존 결과 93.3%가 재현되는지 확인했습니다.
2. 별도 구현 `whitening_compare.py`에서 notch 필터 뒤, 50–300 Hz bandpass 앞에 spectral whitening을 추가했습니다.
3. 전극 60개, 동작 90회, 특징 480개, LDA 분류기와 학습·시험 분할을 동일하게 유지해 두 방법을 비교했습니다.
4. 정확도, 정답을 섞은 대조 검사, 동작별 예측, 주파수별 신호 분포를 저장했습니다.

Whitening은 신호의 주파수별 크기 쏠림을 줄이는 처리입니다. 필터 설정에는 최초 동작 지시 전인 2.00–10.08초의 신호를 사용했습니다.

비교할 때마다 동작 81개로 학습하고, 남겨 둔 9개로 시험했습니다. 10묶음을 돌아가며 시험하는 전체 과정을 10회 반복했습니다. 각 분할에서 분류기를 새로 학습했습니다.

## 평균 정확도 그림

![Whitening 전후 평균 정확도](results/whitening/accuracy_comparison.png)

왼쪽은 기존 방법, 오른쪽은 whitening을 추가한 방법입니다. 막대 높이가 평균 정확도이고, 검은 선은 시험 묶음에 따른 점수 차이를 나타냅니다. 검은 선은 신뢰구간이 아닙니다.

정답을 섞은 검사에서는 약 33%가 나왔습니다. 세 동작을 무작위로 추측할 때의 기준도 약 33%입니다.

## 신호는 어떻게 달라졌나

![Whitening 전후 주파수별 신호 분포](results/whitening/whitening_spectrum.png)

가로축은 **주파수**, 세로축은 **정규화한 파워**입니다. 전극 60개의 중앙값을 그렸습니다.

파란색은 기존 처리 후 신호이고, 주황색은 whitening을 추가한 신호입니다. 주황색 곡선이 더 평평해져 주파수별 크기의 쏠림이 줄어든 것을 볼 수 있습니다.

각 조건을 자기 자신의 평균 파워로 정규화했으므로, 곡선 높이는 주파수별 분포 모양을 비교하는 데 사용합니다. 50 Hz 간격의 골짜기는 notch 필터의 흔적입니다.

## 어떤 동작을 맞혔나

![첫 번째 10-fold 평가의 동작별 예측 비교](results/whitening/confusion_comparison.png)

세로축은 실제 동작, 가로축은 예측한 동작입니다. 대각선의 숫자가 맞힌 횟수입니다.

이 그림은 **첫 번째 10-fold 평가**를 보여줍니다. 기존 방법은 90개 중 84개, whitening 추가 방법은 87개를 맞혔습니다. 각 동작의 정답 수가 30개 중 28개에서 29개로 바뀌었습니다.

이 한 차례의 결과와, 맨 위 표의 10회 반복 평균 96.9%는 구별해서 읽습니다.

## 어디까지 확인한 결과인가
